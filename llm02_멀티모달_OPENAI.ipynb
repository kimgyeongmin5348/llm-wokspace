```python
fp=open("data/audio/베트남여행.mp3","rb")

transcription = client.audio.transcriptions.create(model="whisper-1", file=fp)
print(transcription.text)

summary = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a summarizer pro."},
        {"role": "user", "content": f"Summarize the following text: \n{transcription.text}"}
    ]
)
print(summary.choices[0].message.content)
```

```python
def transcribe_audio(file_path):
    fp = open(file_path, "rb")
    transcription = client.audio.transcriptions.create(model="whisper-1", file=fp)

    fp.close()
    return transcription.text

def transcribe_and_summarize(transcription_text):
    summary = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a summarizer pro."},
            {"role": "user", "content": f"다음 내용을 간단히 요약해줘t: \n{transcription_text}"}
        ]
    )
    return summary.choices[0].message.content

transcription_text = transcribe_audio("data/audio/베트남여행.mp3")
print("오디오로부터 출력된 텍스트 :", transcription_text)

```

```python
import os

response = client.audio.speech.create(input='안녕하세요 오늘 날씨가 참 덥네요', model='tts-1', voice='alloy')

print(response)

os.makedirs('output', exist_ok=True)
with open('output/out.mp3', 'wb') as fp:
    fp.write(response.read())
```

```python
import os
from IPython.display import Audio, display

prompt = input('prompt: ')

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

resp = response.choices[0].message.content
print('gpt-4o-mini response: ', resp)

speech_response = client.audio.speech.create(
    model="tts-1",
    voice="echo",
    instructions="밝고 장난스럽고 어린아이처럼 귀엽게 말해줘.",
    input=prompt
)

os.makedirs('output', exist_ok=True)
with open('output/chat_response.mp3', 'wb') as fp:
    fp.write(speech_response.read())

display(Audio('output/chat_response.mp3', autoplay=True))
```

```python
import base64
import os

response = client.images.generate(
    prompt="A muscular first penguin.",
    model="gpt-image-1",
    size="1024x1024",
    quality="high"
)

image_data = base64.b64decode(response.data[0].b64_json)
os.makedirs("output", exist_ok=True)
with open("output/first_penguin.png", "wb") as f:
    f.write(image_data)

print("Image saved as output/first_penguin.png")
```
